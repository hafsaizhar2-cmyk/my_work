**Q1.** Stop sending emails after 3 users (for loop and break)

```python
users = ["Ali", "Sara", "John", "Zara", "Mike"]
limit = 3
count = 0

for user in users:
    if count == limit:
        print("Limit reached. Stop sending emails.")
        break
    print("Email sent to", user)
    count += 3
    
```

**Q2.** Stop monitoring temperature if critical temperature reached (for loop and break)

```python
temps = [22, 25, 30, 45, 28]
critical = 40

for t in temps:
    if t >= critical:
        print("Critical temperature reached! Stop monitoring.")
        break
    print("Temperature:", t)
```

**Q3.** Calculate total price of items in a shopping cart (for loop)

```python
cart = [100, 250, 50, 300]
total = 0

for price in cart:
    total += price

print("Total Price:", total)
```

**Q4.** Skip customers with unpaid bills (for loop and continue)

```python
customers = [
    {"name": "Ali", "paid": True},
    {"name": "Sara", "paid": False},
    {"name": "John", "paid": True}
]

for customer in customers:
    if not customer["paid"]:
        continue
    print("Access granted to:", customer["name"])
```

**Q5.** Send personalized emails to a list of users (for loop)

```python
users = ["Ali", "Sara", "John"]

for user in users:
    print(f"Hello {user}, your subscription is active!")
```

**Q6.** Extract domain names from email list (for loop)

```python
emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]
domains = []

for email in emails:
    domains.append(email.split("@")[1])

print("domains:", domains)
```

**Q7.** New list of numbers divisible by 3 or 5 (list comprehension)

```python
numbers = list(range(1, 31))
result = [n for n in numbers if n % 3 == 0 or n % 5 == 0]

print("Original:", numbers)
print("Divisible by 3 or 5:", result)
```

**Q8.** Multiplication table using a nested for loop

```python
numbers = [2, 3, 4]
multipliers = [1, 2, 3]

for n in numbers:
    for m in multipliers:
        print(f"{n} x {m} = {n * m}")
```

**Q9.** Keep asking for a password until it is correct (while loop)

```python
correct_password = "python123"

password = input("python123")
while password != correct_password
    print("Incorrect password. Try again.\n")
    password = input("python123")

print("Access granted!")
```

**Q10.** Count down from 5 to 1, then display "Time's up!" (while loop)

```python
count = 5

while count >= 1:
    print(count)
    count -= 1

```
